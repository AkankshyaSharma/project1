#outliers detection

```python
num_df = df.select_dtypes(include=np.number)
```

```python
def detect_outliers_iqr(data):
    outlier_info = {}

    for col in data.columns:
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
        outlier_info[col] = len(outliers)

    return outlier_info
```

```python
outlier_counts = detect_outliers_iqr(num_df)
outlier_counts
```

```python
log_cols = [
    'GDP', 'Population Total', 'CO2 Emissions', 'Energy Usage',
    'Health Exp/Capita', 'Tourism Inbound', 'Tourism Outbound'
]

for col in log_cols:
    if col in num_df.columns:
        num_df[col] = np.log1p(num_df[col])
```

Why NOT cap these?

These define development differences

Capping would flatten rich vs poor countries

Log keeps ranking but reduces dominance

```python
cap_cols = [
    'Business Tax Rate', 'Days to Start Business',
    'Hours to do Tax', 'Lending Interest',
    'Health Exp % GDP', 'Infant Mortality Rate'
]

for col in cap_cols:
    Q1 = num_df[col].quantile(0.25)
    Q3 = num_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    num_df[col] = num_df[col].clip(lower, upper)
```

Why cap instead of log?

Values are bounded in reality

Extremes are often distortions

We want to limit influence, not rescale hierarchy

Model Evaluation and Hyperparameter Tuning

1️ K-MEANS CLUSTERING

K-Means groups countries by minimizing distance between similar development indicators.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(6,4))
plt.plot(range(2,11), wcss, marker='o')
plt.title("Elbow Method for K-Means")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()
```

Final K-Means Model

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans_labels = kmeans.fit_predict(X)

X['KMeans_Cluster'] = kmeans_labels
 
 
```

Evaluation

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score

print("Silhouette Score:", silhouette_score(X.drop(columns='KMeans_Cluster'), kmeans_labels))
print("Davies-Bouldin Index:", davies_bouldin_score(X.drop(columns='KMeans_Cluster'), kmeans_labels))
```

CLUSTER VISUALIZATION

PCA setup

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
```

K-Means Visualization

```python
plt.figure(figsize=(6,5))
sns.scatterplot(
    x=X_pca[:,0],
    y=X_pca[:,1],
    hue=kmeans_labels,
    palette='Set2'
)
plt.title("K-Means Clusters (PCA)")
plt.show()
```

2️ HIERARCHICAL CLUSTERING

Hierarchical clustering builds a tree-like structure showing how countries merge based on similarity.

Dendrogram

```python
from scipy.cluster.hierarchy import dendrogram, linkage

linked = linkage(X.drop(columns='KMeans_Cluster'), method='ward')

plt.figure(figsize=(10,5))
dendrogram(linked)
plt.title("Hierarchical Clustering Dendrogram")
plt.show()
```

Agglomerative Clustering

```python
from sklearn.cluster import AgglomerativeClustering

hc = AgglomerativeClustering(n_clusters=3)
hc_labels = hc.fit_predict(X.drop(columns='KMeans_Cluster'))

X['Hierarchical_Cluster'] = hc_labels
```

Evaluation

```python
print("Silhouette Score:", silhouette_score(X.drop(columns=['KMeans_Cluster','Hierarchical_Cluster']), hc_labels))
print("Davies-Bouldin Index:", davies_bouldin_score(X.drop(columns=['KMeans_Cluster','Hierarchical_Cluster']), hc_labels))
```

3️ DBSCAN (Noise Detection)

DBSCAN identifies dense regions and treats outliers as noise.

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=1.2, min_samples=5)
dbscan_labels = dbscan.fit_predict(X.drop(columns=['KMeans_Cluster','Hierarchical_Cluster']))

X['DBSCAN_Cluster'] = dbscan_labels
```

Evaluation

```python
mask = dbscan_labels != -1

print("Silhouette Score:", silhouette_score(
    X.drop(columns=['KMeans_Cluster','Hierarchical_Cluster','DBSCAN_Cluster'])[mask],
    dbscan_labels[mask]
))
```

DBSCAN Visualization

```python
plt.figure(figsize=(6,5))
sns.scatterplot(
    x=X_pca[:,0],
    y=X_pca[:,1],
    hue=dbscan_labels,
    palette='tab10'
)
plt.title("DBSCAN Clusters (PCA)")
plt.show()
```

4 GMM (Gaussian Mixture Model)

GMM performs probabilistic (soft) clustering.

```python
from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=3, random_state=42)
gmm_labels = gmm.fit_predict(X.drop(columns=['KMeans_Cluster','Hierarchical_Cluster','DBSCAN_Cluster']))

X['GMM_Cluster'] = gmm_labels
```

Evaluation

```python
print("Silhouette Score:", silhouette_score(
    X.drop(columns=['KMeans_Cluster','Hierarchical_Cluster','DBSCAN_Cluster','GMM_Cluster']),
    gmm_labels
))
```

GMM Visualization

```python
plt.figure(figsize=(6,5))
sns.scatterplot(
    x=X_pca[:,0],
    y=X_pca[:,1],
    hue=gmm_labels,
    palette='Set1'
)
plt.title("GMM Clusters (PCA)")
plt.show()
```

5 MEAN SHIFT CLUSTERING


Mean Shift is a density-based clustering algorithm that automatically detects the number of clusters without requiring prior specification of cluster count.

```python
from sklearn.cluster import MeanShift

ms = MeanShift()
ms_labels = ms.fit_predict(
    X.drop(columns=[
        'KMeans_Cluster',
        'Hierarchical_Cluster',
        'DBSCAN_Cluster',
        'GMM_Cluster'
    ])
)

X['MeanShift_Cluster'] = ms_labels
```

Evaluation — Silhouette Score

```python
print(
    "Silhouette Score:",
    silhouette_score(
        X.drop(columns=[
            'KMeans_Cluster',
            'Hierarchical_Cluster',
            'DBSCAN_Cluster',
            'GMM_Cluster',
            'MeanShift_Cluster'
        ]),
        ms_labels
    )
)
```

Visualization — PCA Scatter Plot

```python
plt.figure(figsize=(6,5))
sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=ms_labels,
    palette='Dark2'
)
plt.title("Mean Shift Clusters (PCA)")
plt.show()
```
