#outliers detection

```python
num_df = df.select_dtypes(include=np.number)
```

```python
def detect_outliers_iqr(data):
    outlier_info = {}

    for col in data.columns:
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outliers = data[(data[col] < lower_bound) | (data[col] > upper_bound)]
        outlier_info[col] = len(outliers)

    return outlier_info
```

```python
outlier_counts = detect_outliers_iqr(num_df)
outlier_counts
```

```python
log_cols = [
    'GDP', 'Population Total', 'CO2 Emissions', 'Energy Usage',
    'Health Exp/Capita', 'Tourism Inbound', 'Tourism Outbound'
]

for col in log_cols:
    if col in num_df.columns:
        num_df[col] = np.log1p(num_df[col])
```

Why NOT cap these?

These define development differences

Capping would flatten rich vs poor countries

Log keeps ranking but reduces dominance

```python
cap_cols = [
    'Business Tax Rate', 'Days to Start Business',
    'Hours to do Tax', 'Lending Interest',
    'Health Exp % GDP', 'Infant Mortality Rate'
]

for col in cap_cols:
    Q1 = num_df[col].quantile(0.25)
    Q3 = num_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    num_df[col] = num_df[col].clip(lower, upper)
```

Why cap instead of log?

Values are bounded in reality

Extremes are often distortions

We want to limit influence, not rescale hierarchy

RobustScaler was used because it scales features using median and interquartile range, making it robust to outliers commonly present in global development indicators, unlike StandardScaler and MinMaxScaler which are highly sensitive to extreme values.

```python
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
#KMeans Clustering
kmeans=KMeans(n_clusters=3,random_state=0)
kmeans.fit(scaled_data)
```

```python
kmeans.inertia_
```

```python
wcss = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_data)
    wcss.append(kmeans.inertia_)
```

```python
import matplotlib.pyplot as plt

plt.plot(range(2, 11), wcss)
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

```python
from sklearn.cluster import KMeans
clusters_new = KMeans(n_clusters=4, random_state=42)
labels=clusters_new.fit_predict(scaled_data)
labels
```

```python
#DBSCAN
dbscan = DBSCAN(eps=1.5, min_samples=30)
dbscan_labels = dbscan.fit_predict(X)
dbscan_labels
```

-1 are the noise points (outliers)

```python
#Compute k-distance
from sklearn.neighbors import NearestNeighbors
import numpy as np

k = 20  # same as min_samples

neighbors = NearestNeighbors(n_neighbors=k)
neighbors_fit = neighbors.fit(scaled_data)

distances, indices = neighbors_fit.kneighbors(scaled_data)
```

```python
#sort the distance
distances = np.sort(distances[:, -1])
```

```python
#plot K-distance graph
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.plot(distances)
plt.xlabel("Data Points")
plt.ylabel("k-distance")
plt.title("k-Distance Graph for DBSCAN")
plt.show()
```

The eps parameter was selected using the k-distance graph. The elbow point, where the distance curve shows a sharp change in slope, occurs around 3.3. This value represents the optimal neighborhood radius before noise points dominate.

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=3.3, min_samples=20)
dbscan_labels = dbscan.fit_predict(scaled_data)
dbscan_labels
```

#Model evaluation

```python
# K-Means evaluation
score = silhouette_score(scaled_data, labels)
print("Silhouette Score:", score)
```

```python
#Remove noise points
mask = dbscan_labels != -1


#DBSCAN evaluation
score = silhouette_score(scaled_data, dbscan_labels)
print("DBSCAN Silhouette Score:", score)
```

K-Means clustering achieved a silhouette score of 0.17, indicating weak cluster separation. DBSCAN, a density-based algorithm, achieved a higher silhouette score of 0.34 by identifying dense regions and treating outliers as noise. Given the complex and overlapping nature of global development indicators, DBSCAN provided more meaningful and robust clustering results and was therefore selected as the final model.
